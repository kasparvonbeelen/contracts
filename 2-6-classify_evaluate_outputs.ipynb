```python
import pandas as pd
from sklearn.metrics import classification_report, confusion_matrix
```

```python
clause_type, clause_col =  'arbitration', 'arb_clause'  # 'arbitration', 'arb_clause' | 'class_waiver', 'class_clause'
pred = pd.read_csv(f'results/{clause_type}_heatmap_data.csv')
```

```python
pred.set_index('platform', inplace=True)
pred = pred.unstack().reset_index().rename(columns={0: 'pred', 'level_0': 'year'})
pred = pred[pred['pred'] > -1]

pred['year'] = pred.year.astype('int64')
pred['pred'] = pred['pred'].astype('int64')
```

```python
gt = pd.read_csv(f'annotations/evalutation_yearly/{clause_type}_binary.csv')
gt['platform'] = gt['platform'].str.lower()
```

```python
data = gt.merge(pred, on=['platform', 'year'], how='inner')
```

```python
print(classification_report(data[clause_col], data['pred']))
```
